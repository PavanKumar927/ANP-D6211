```python
import random
import pandas as pd 
from faker import Faker



# active my fake fellow to make ready for generating the random values
fake = Faker()
total_records = 10000

data = [] 


for _ in range (total_records):
    record = {
        "Region":random.choice(["East", "Central", "West","South"]),	
        "Manager":fake.name(),	
        "SalesMan":fake.name(),	
        "Email":fake.email(),
        "OrderDate": fake.date_between(start_date="-5y",end_date="today"),
        "Item":random.choice(["Television","Home Theater","Cell Phone","Desk","Video Games"]),	
        "Units":round(random.uniform(1, 100),0),	
        "Unit_price":round(random.uniform(50, 1200),1),	
        "Sale_amt":round(random.uniform(250, 140000),2)
    }
    data.append(record)
df = pd.DataFrame(data)
df.to_excel("dataRandom.xlsx",index = False)


# print(data)    




```
